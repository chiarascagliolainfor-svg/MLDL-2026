# Proprietà e caratteristiche
.index etichetta indici , dtype tipologia di dato , size numero totale di elementi , empty se è vuoto ci restituisce un bool

## slicing
.loc[] seleziona elemento di quell'indice ( nome dell'indice )
.iloc[] stessa cosa con indice numerico

```python
prezzi_prodotti = pd.Series([3,4,5,6], index= ['Mouse', 'Tastiera', 'Cuffie', 'Telefono'], name = 'Prodotti')

print( 'prezzo mouse ', prezzi_prodotti.loc['Mouse'])
print(' prezzo tastiera e mouse \n ', prezzi_prodotti.loc[['Tastiera', 'Mouse']])
```

```python
print('primo elemento ', prezzi_prodotti.iloc[0])
```

# slicing
.loc[inzio: fine ] fine è inclusa

.iloc[inizio: fine] fine è esclusa

vale anche il passo [inizio: fine: passo ]

```python
# prezzi primi 3 prodotti
print('prezzi primi 3 prodotti')
print( prezzi_prodotti.iloc[0:3] )
```

```python
# prezzi primi 4 prodotti alternati (il primo e il terzo)
print('prezzi primi 3 prodotti')
print( prezzi_prodotti.iloc[0:4:2] )
```

```python
print( prezzi_prodotti.loc['Mouse': 'Cuffie'] )
```

# condizioni
AND--> & OR--> | NOT--> ~  .between() ( estremi inclusi in between) .isin()

```python
print('prezzi>=15 e =<20')
print(prezzi_reali[(prezzi_reali >=15) & (prezzi_reali <=20)])
```

```python
print(prezzi_reali[prezzi_reali.between(15,20)])
```

```python
# voglio capire se la mia serie contiene mouse e tastiera
```

# gestione valori nulli
.isna() --> verifica i valori mancanti ( anche questo ritorna una mask booleana)

.notna() --> verifica i valori non nulli

```python
valutazioni = pd.Series([4.2 , None, 5, None], index=['PC','Telefono', 'Mouse', 'Cuffie'])
print('maschera')
print( valutazioni.isna() )
```

```python
## voglio i valori non nulli ?
print(valutazioni[valutazioni.notna()])

print(valutazioni[valutazioni.notna()].size)
```

# gestione valori nulli
.dropna()  elimina i valori nulli
.fillna() posso mettere dentro i valori nulli un valore che voglio !

```python
valutazioni_nuove = valutazioni.copy()
print( valutazioni_nuove.dropna())
media = valutazioni_nuove.mean()
print('valutazioni con filling media', valutazioni_nuove.fillna(media))
```

# ordinamento valori
.sort_values() oridana per valore

.sort_index() ordina per indice (ascending = False)

```python
print('ordinamento per valore ')
print(valutazioni.sort_values())
print('ordinamento per indice')
print(valutazioni.sort_index(ascending=False))
```

## selezionare righe e colonne per il nostro df
- .loc[] etichetta di riga e di colonna
- .iloc[] indice riga e colonna (estremo superiore escluso)
sintassi : .loc[righe inizio: fine, colonne inizio: fine]

```python
print('prime 2 righe e prime 3 colonne con iloc')
print(df_prodotti_finale.iloc[:2,:3])
print( 'righe da 1 a 3 e colonne prezzo e nome:')
print(df_prodotti_finale.loc[1:3,['prezzo','nome']])
```
